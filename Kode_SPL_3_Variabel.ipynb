```python
# @title
# PROGRAM SPLTV METODE ELIMINASI GAUSS
# Dapat dijalankan langsung di Google Colab

def eliminasi_gauss(matriks):
    """
    Menyelesaikan SPLTV dengan metode Eliminasi Gauss.
    Matriks berukuran 3 x 4 (3 koefisien dan 1 konstanta).
    """

    n = 3

    # Membuat salinan matriks agar data asli tidak berubah
    A = [baris[:] for baris in matriks]

    # Proses eliminasi Gauss
    for i in range(n):
        # Jika elemen diagonal nol, cari baris lain untuk ditukar
        if abs(A[i][i]) < 1e-12:
            baris_pengganti = None
            for k in range(i + 1, n):
                if abs(A[k][i]) >= 1e-12:
                    baris_pengganti = k
                    break

            if baris_pengganti is None:
                raise ValueError(
                    "Sistem tidak memiliki solusi tunggal. "
                    "Periksa kembali koefisien yang dimasukkan."
                )

            A[i], A[baris_pengganti] = A[baris_pengganti], A[i]

        # Menghilangkan variabel pada baris di bawah pivot
        for j in range(i + 1, n):
            faktor = A[j][i] / A[i][i]
            for k in range(i, n + 1):
                A[j][k] = A[j][k] - faktor * A[i][k]

    # Substitusi mundur
    hasil = [0.0, 0.0, 0.0]

    for i in range(n - 1, -1, -1):
        jumlah = sum(A[i][j] * hasil[j] for j in range(i + 1, n))

        if abs(A[i][i]) < 1e-12:
            raise ValueError(
                "Sistem tidak memiliki solusi tunggal. "
                "Periksa kembali koefisien yang dimasukkan."
            )

        hasil[i] = (A[i][n] - jumlah) / A[i][i]

    return hasil, A


print("PROGRAM PENYELESAIAN SPLTV - METODE ELIMINASI GAUSS")
print("Masukkan koefisien untuk persamaan:")
print("a1*x + b1*y + c1*z = d1")
print("a2*x + b2*y + c2*z = d2")
print("a3*x + b3*y + c3*z = d3")

matriks = []

# Input tiga persamaan
for i in range(3):
    print(f"\nPersamaan ke-{i + 1}")
    a = float(input("Koefisien x: "))
    b = float(input("Koefisien y: "))
    c = float(input("Koefisien z: "))
    d = float(input("Konstanta: "))

    matriks.append([a, b, c, d])

# Menampilkan matriks augmented
print("\nMatriks augmented:")
for baris in matriks:
    print(baris[:3], "|", baris[3])

# Menghitung dan menampilkan hasil
try:
    hasil, matriks_akhir = eliminasi_gauss(matriks)

    print("\nMatriks setelah eliminasi Gauss:")
    for baris in matriks_akhir:
        print([round(nilai, 4) for nilai in baris[:3]],
              "|", round(baris[3], 4))

    print("\nHASIL PERHITUNGAN:")
    print(f"Nilai x = {hasil[0]:.4f}")
    print(f"Nilai y = {hasil[1]:.4f}")
    print(f"Nilai z = {hasil[2]:.4f}")

except ValueError as error:
    print("\nTidak dapat menyelesaikan SPLTV:", error)



13
```
