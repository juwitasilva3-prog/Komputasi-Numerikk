```python
# ==================================================
# PROGRAM SPLTV METODE ELIMINASI GAUSS
# Dijalankan menggunakan Google Colab
# ==================================================

# Fungsi untuk menyelesaikan SPLTV
def eliminasi_gauss(matriks):

    # Menentukan jumlah variabel, yaitu x, y, dan z
    n = 3

    # Membuat salinan matriks agar data asli tidak berubah
    A = [baris[:] for baris in matriks]

    # PROSES ELIMINASI GAUSS
    for i in range(n):

        # Memeriksa apakah elemen diagonal bernilai nol
        if abs(A[i][i]) < 1e-12:

            # Mencari baris lain untuk ditukar
            baris_pengganti = None

            for k in range(i + 1, n):
                if abs(A[k][i]) >= 1e-12:
                    baris_pengganti = k
                    break

            # Jika tidak ditemukan baris pengganti
            if baris_pengganti is None:
                raise ValueError(
                    "Sistem tidak memiliki solusi tunggal."
                )

            # Menukar posisi baris
            A[i], A[baris_pengganti] = A[baris_pengganti], A[i]

        # Menghilangkan variabel pada baris di bawah pivot
        for j in range(i + 1, n):

            # Menghitung faktor eliminasi
            faktor = A[j][i] / A[i][i]

            # Mengubah nilai setiap kolom
            for k in range(i, n + 1):
                A[j][k] = A[j][k] - faktor * A[i][k]

    # PROSES SUBSTITUSI MUNDUR

    # Menyiapkan tempat untuk menyimpan nilai x, y, dan z
    hasil = [0.0, 0.0, 0.0]

    # Menghitung nilai variabel dari bawah ke atas
    for i in range(n - 1, -1, -1):

        # Menghitung jumlah hasil perkalian variabel
        jumlah = sum(
            A[i][j] * hasil[j]
            for j in range(i + 1, n)
        )

        # Memeriksa apakah elemen diagonal bernilai nol
        if abs(A[i][i]) < 1e-12:
            raise ValueError(
                "Sistem tidak memiliki solusi tunggal."
            )

        # Menghitung nilai variabel
        hasil[i] = (A[i][n] - jumlah) / A[i][i]

    # Mengembalikan hasil perhitungan dan matriks
    return hasil, A


# ==================================================
# BAGIAN INPUT DATA
# ==================================================

print("PROGRAM PENYELESAIAN SPLTV")
print("METODE ELIMINASI GAUSS")

# Menampilkan bentuk umum persamaan
print("\nMasukkan koefisien tiga persamaan:")
print("a1*x + b1*y + c1*z = d1")
print("a2*x + b2*y + c2*z = d2")
print("a3*x + b3*y + c3*z = d3")

# Membuat daftar kosong untuk menyimpan matriks
matriks = []

# Memasukkan data dari tiga persamaan
for i in range(3):

    print(f"\nPersamaan ke-{i + 1}")

    # Input koefisien variabel x
    a = float(input("Koefisien x: "))

    # Input koefisien variabel y
    b = float(input("Koefisien y: "))

    # Input koefisien variabel z
    c = float(input("Koefisien z: "))

    # Input nilai konstanta
    d = float(input("Konstanta: "))

    # Menyimpan data ke dalam matriks
    matriks.append([a, b, c, d])


# ==================================================
# MENAMPILKAN MATRIKS AUGMENTED
# ==================================================

print("\nMATRIKS AUGMENTED:")

# Menampilkan koefisien dan konstanta
for baris in matriks:
    print(baris[:3], "|", baris[3])


# ==================================================
# PROSES PERHITUNGAN DAN OUTPUT
# ==================================================

try:

    # Memanggil fungsi eliminasi Gauss
    hasil, matriks_akhir = eliminasi_gauss(matriks)

    # Menampilkan matriks setelah eliminasi
    print("\nMATRIKS SETELAH ELIMINASI GAUSS:")

    for baris in matriks_akhir:
        print(
            [round(nilai, 4) for nilai in baris[:3]],
            "|",
            round(baris[3], 4)
        )

    # Menampilkan hasil akhir perhitungan
    print("\nHASIL PERHITUNGAN SPLTV:")
    print("Nilai x =", round(hasil[0], 4))
    print("Nilai y =", round(hasil[1], 4))
    print("Nilai z =", round(hasil[2], 4))

# Menangani kesalahan jika sistem tidak memiliki solusi tunggal
except ValueError as error:
    print("\nTerjadi kesalahan:", error)
```

PROGRAM PENYELESAIAN SISTEM PERSAMAAN LINEAR TIGA VARIABEL (SPLTV)

Metode: Eliminasi Gauss

Tujuan: Program ini dibuat untuk menyelesaikan sistem persamaan linear tiga variabel, yaitu x, y, dan z, menggunakan metode Eliminasi Gauss.

Metode ini bekerja dengan mengubah persamaan menjadi matriks augmented, melakukan operasi baris elementer, kemudian mencari nilai variabel menggunakan substitusi mundur.

PENJELASAN PROGRAM

Input Data Pengguna memasukkan koefisien x, y, z dan konstanta dari tiga persamaan.

Matriks Augmented Koefisien dan konstanta digabungkan menjadi satu matriks untuk mempermudah perhitungan.

Eliminasi Gauss Program mengubah elemen di bawah diagonal utama menjadi nol menggunakan operasi baris elementer.

Substitusi Mundur Program menghitung nilai variabel mulai dari variabel terakhir sampai variabel pertama.

Output Program menampilkan hasil nilai x, y, dan z.
